## DB 저장

```python
def save_to_postgresql(df, db_url, table_name):
    df_save = df.copy()

    for col in df_save.columns:
        if pd.api.types.is_string_dtype(df_save[col]):
            df_save[col] = df_save[col].astype(str)
            
    engine = create_engine(db_url)

    with engine.connect() as conn:
        version = conn.execute(text('SELECT version();')).fetchone()[0]
        print('PostgreSQL 연결 성공')
        print(version[:80] + '...')

    df_save.to_sql(
        name=table_name,
        con=engine,
        if_exists='replace',
        index=False,
        chunksize=1000,
        method='multi'
    )

    with engine.connect() as conn:
        saved_count = conn.execute(text(f'SELECT COUNT(*) FROM {table_name};')).fetchone()[0]
    print(f'저장 완료 : {saved_count}행')
    print(f'DB : busapidb / table : {table_name}')
```

```python
save_to_postgresql(df_route, DB_URL, TABLE_NAME)
```

## CSV 저장

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)

df_route.to_csv(OUTPUT_PATH, encoding='utf-8-sig', index=False)
print('CSV 저장 완료')
```

## 여러 정류소 반복 조회

```python
# 정류소 선택
node_dict = {
    'DGB7041053900':'죽전119안전센터건너',
    'DGB7031004200':'광장타운앞',
    'DGB7051013500':'현충삼거리2',
    'DGB7051003200':'대명초등학교앞',
    'DGB7111040300':'스파밸리1'
    }
```

```python
OUTPUT_PATH_2 = os.path.join(OUTPUT_DIR, 'bus_route_2.csv')
print(f'CSV 파일 저장 경로 : {OUTPUT_PATH_2}')
```

```python
all_data = []
for nodeid, nodenm in node_dict.items():
    response2 = requests.get(BASE_URL,
                            params={
                                'serviceKey':API_KEY,
                                'pageNo':1,
                                'numOfRows':10,
                                '_type':'json',
                                'cityCode':CITY_CODE,
                                'nodeid':nodeid
                            })
    data2 = response2.json()
    data_items = data2['response']['body']['items']['item']
    all_data.extend(data_items)
all_data
```

```python
all_df_list = []
for nodeid, nodenm in node_dict.items():
    response2 = requests.get(BASE_URL,
                            params={
                                'serviceKey':API_KEY,
                                'pageNo':1,
                                'numOfRows':10,
                                '_type':'json',
                                'cityCode':CITY_CODE,
                                'nodeid':nodeid
                            })
    data2 = response2.json()
    data_items = data2['response']['body']['items']['item']
    
    df_temp = pd.DataFrame(data_items)

    df_temp['정류소ID'] = nodeid
    df_temp['정류소명'] = nodenm

    all_df_list.append(df_temp)
all_df = pd.concat(all_df_list, ignore_index=True)
all_df
```

```python
COLUMN_NAME = {
    'endnodenm':'종점',
    'routeid':'노선ID',
    'routeno':'노선번호',
    'routetp':'노선유형',
    'startnodenm':'기점'
}
df_raw_2 = all_df.rename(columns=COLUMN_NAME)
df_raw_2.head()
```

```python
df2 = df_raw_2[['노선ID', '노선번호', '정류소ID', '정류소명']]
df2.head()
```

```python
save_to_postgresql(df2, DB_URL, 'route_by_stop_2')
```
