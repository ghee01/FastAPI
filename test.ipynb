```python
import requests
import os
from dotenv import load_dotenv
import pandas as pd
```

```python
load_dotenv()
API_KEY = os.getenv('TAGO_API_KEY')

if API_KEY:
    print(f'API 키 로드 완료: {API_KEY[:6]}...{API_KEY[-4:]}')
else:
    print('env 파일에 API 키를 설정하세요')
```

```python
url = 'https://apis.data.go.kr/B553077/api/open/sdsc2/storeListInUpjong'
params = {
    'ServiceKey':API_KEY,
    'pageNo':1,
    'numOfRows':10,
    'divId':'indsSclsCd',
    'key':'I21201',
    'type':'json'
}
response = requests.get(url, params)
response.status_code
```

```python
data = response.json()
data
```

```python
data_body = data['body']
data_body
```

```python
data_item = data_body['items']
data_item
```

```python
df = pd.DataFrame(data_item)
df
```
