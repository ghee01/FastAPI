```python
import pandas as pd
```

## 화재 발생 현황 데이터 전처리

```python
fire_raw = pd.read_csv(
    'datafile/화재발생_서울.csv',
    encoding='utf-8-sig',
    skiprows=[1]
)
fire_raw.head()
```

```python
fire_raw = fire_raw.rename(columns={
    '동별(1)':'시도',
    '동별(2)':'구',
    '시점':'연도',
    '발생 (건)':'화재발생건수',
    '피해액 (천원)':'피해액_천원',
    '인명피해 (명)':'인명피해'
})

fire = fire_raw[fire_raw['구'] == '소계'][['연도', '화재발생건수', '피해액_천원']].copy()
fire.insert(0, '지역', '서울')
```

```python
fire.head()
```

## 소방 장비 데이터 전처리

```python
equip_raw = pd.read_csv(
    'datafile/소방장비_서울.csv',
    encoding='utf-8-sig',
    skiprows=[1, 2]
)
equip_raw.head()
```

```python
equip_raw = equip_raw.rename(columns={
    '소방서별(1)':'시도',
    '소방서별(2)':'소방서',
    '시점':'연도',
    '합계':'소방장비대수'
})

equip = equip_raw[equip_raw['소방서'] == '소계'][['연도', '소방장비대수']].copy()
equip.insert(0, '지역', '서울')
```

```python
equip.head()
```

## df, equip 테이블 합치기

```python
df_seoul = pd.merge(
    equip, fire,
    on=['지역', '연도'],
    how='outer'
)
df_seoul
```

```python
df_seoul.to_csv('seoul_data.csv', encoding='utf-8-sig', index=False)
```
