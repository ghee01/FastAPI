```python
import pandas as pd
```

## 서울

### 화재 발생

```python
fire_raw = pd.read_csv(
    'datafile/화재발생_서울.csv',
    encoding='utf-8-sig',
    skiprows=[1]
)
fire_raw.head()
```

```python
fire_raw = fire_raw.rename(columns={
    '동별(1)':'시도',
    '동별(2)':'구',
    '시점':'연도',
    '발생 (건)':'화재발생건수',
    '피해액 (천원)':'피해액_천원',
    '사망 (명)':'사망',
    '부상 (명)':'부상',
    '인명구조 (명)':'구조인원'
})

fire = fire_raw[fire_raw['구'] == '소계'][['연도', '화재발생건수', '피해액_천원', '사망', '부상', '구조인원']].copy()
fire.insert(0, '지역', '서울')
```

```python
fire.head()
```

### 소방 장비

```python
equip_raw = pd.read_csv(
    'datafile/소방장비_서울.csv',
    encoding='utf-8-sig',
    skiprows=[1, 2]
)
equip_raw.head()
```

```python
equip_raw = equip_raw.rename(columns={
    '소방서별(1)':'시도',
    '소방서별(2)':'소방서',
    '시점':'연도',
    '합계':'소방장비대수'
})

equip = equip_raw[equip_raw['소방서'] == '소계'][['연도', '소방장비대수']].copy()
equip.insert(0, '지역', '서울')
```

```python
equip.head()
```

### 테이블 병합

```python
def merge_table(fire_table, equip_table):
    df = pd.merge(
    equip_table, fire_table,
    on=['지역', '연도'],
    how='outer'
    )
    return df
```

```python
df_seoul = merge_table(fire, equip)
df_seoul
```
