```python
import pandas as pd
```

```python
df_19 = pd.read_csv('경찰청_범죄 발생 지역별 통계_20191231.csv', encoding='cp949')
df_19.head()
```

```python
region_cols = ['서울', '부산', '대구', '인천', '대전', '울산', '세종']
years = {
    2019:'20191231',
    2020:'20201231',
    2021:'20211231',
    2022:'20221231',
}
```

```python
dfs = []
for year, suffix in years.items():
    df = pd.read_csv(f'경찰청_범죄 발생 지역별 통계_{suffix}.csv', encoding='cp949')

    long_df = df.melt(
        value_vars=region_cols,
        var_name='지역',
        value_name='범죄발생수'
    )

    total = long_df.groupby(by='지역', as_index=False)['범죄발생수'].sum()
    total['연도'] = year

    dfs.append(total)

dfs
```

```python
crime_long = pd.concat(dfs, ignore_index=True)
crime_long = crime_long[['지역', '연도', '범죄발생수']]
crime_long
```

```python
crime_long = crime_long.sort_values(['지역', '연도']).reset_index(drop=True)
crime_long
```

```python
crime_long.to_csv('crime_by_region_year.csv', encoding='utf-8-sig', index=False)
```
